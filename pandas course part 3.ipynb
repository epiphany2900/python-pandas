```python
import pandas as pd
import numpy as np
```

```python
data = {
    "Name": ["Fatin", "Nabila", "Farah", "Hannani"],
    "Age": [25, None, 30, np.nan],
    "City": ["NY", "LA", None, "Chicago"]
}
```

```python
df = pd.DataFrame(data)
df
```

```python
df.isnull().sum()
```

```python
df.isnull().sum().sum()
```

```python
df_filled = df.fillna({"Age": df['Age'].mean(), "City": 'Unknown'})
```

```python
df_filled
```

```python
df_filled.isnull().sum().sum()
```

```python
df
```

```python
df_dropped = df.dropna()
```

```python
df_dropped
```

```python
df.dtypes
```
