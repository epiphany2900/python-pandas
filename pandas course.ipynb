```python
data = [
    ["Alice", 24],
    ["Bob", 30],
    ["Charlie", 19]
]
```

```python
s1 = pd.Series([1,2,3], index=["a","b","c"])
s2 = pd.Series([10,20,30], index=["a","b","c"])
```

```python
df = pd.DataFrame({"col1": s1, "col2": s2})
df
```

```python
df = pd.DataFrame(columns=["name", "age"])
```

```python
df = pd.DataFrame({
    "date": ["08-10-2023", "08-10-2024", "08-10-2025", "08-10-2026", "08-10-2027"],
    "product": ["tea", "coffee", "tea", "coffee", "tea"],
    "quantity": [4,2,3,5,10],
    "price": [40, 40, 30, 100, 10]
})
df
```

```python
df.to_csv("sales_example.csv", index=False)
```

```python
df.to_excel("sales_example.xlsx", index=False)
```

```python
df.to_json("sales_example.json", orient="records", lines=False)
```

```python
csv_df = pd.read_csv("sales_example.csv", parse_dates=["date"])
csv_df
```
