```python
from pydoc import describe

import pandas as pd
```

```python
data = {
    "Product": ["A","A","A","B","B","C"],
    "Category": ["Food","Food","Food","Drinks","Drinks", "Snacks"],
    "Sales": [10,15,7,20,22,5],
    "Store": ["X","Y","X","Y","X","Y"]
}
```

```python
df = pd.DataFrame(data)
df
```

```python
df.groupby("Product")["Sales"].sum()
```

```python
df.groupby("Product")["Sales"].mean()
```

```python
df.groupby("Product")["Sales"].median()
```

```python
df.groupby("Store")["Sales"].sum()
```

```python
df.groupby("Category")["Sales"].max().reset_index()
```

```python
df
```

```python
df.groupby(["Product", "Store"])["Sales"].mean().reset_index()
```

```python
df
```

```python
df.groupby("Category").agg({
    "Sales": ["sum","mean","count"]
})
```

```python
df
```

```python
df.pivot_table(
    values= "Sales", # what you want to summarize
    index = "Product", # this is the rows
    columns = "Store", # new columns created
    aggfunc= "sum" # calculation to apply inside each cell
)
```

```python
df
```

```python
df.pivot_table(
    values= "Sales",
    index= "Category",
    columns= "Store",
    aggfunc= ["sum", "mean"]
)
```

```python
pd.crosstab(df["Product"], df["Store"])
```

```python
pd.crosstab(df["Product"], df["Store"], margins=True)
```

```python
pd.crosstab(df["Product"], df["Store"], values= df["Sales"], aggfunc="sum")
```
