```python
#binning/discretization
# 0 to 100
# 0 to 17 child
# 18 to 64 adult
# 65+ senior

# poor, fair, good, excellent
```

```python
ages = [15,22,37,45,72,16]
```

```python
df = pd.DataFrame({"ages": ages})
df
```

```python
bins = [0,18,65,120]
```

```python
labels = ["child", "adult", "senior"]
```

```python
df["group"] = pd.cut(df["ages"], bins=bins, labels=labels)
df
```

```python
df = pd.DataFrame({"x": [1.2, 3.5, 5.8,9.1]})
df
```

```python
df["bin"] = pd.cut(df["x"], bins=3)
df
```

```python
data = {
    "price": [100,200,150],
    "quantity": [2,3,4],
    "date": ["2024 01 01", "2024 01 02", "2024 01 03"]
}
```

```python
df = pd.DataFrame(data)
```

```python
df
```

```python
df["total"] = df["price"] * df["quantity"]
```

```python
df
```

```python
df["date"] = pd.to_datetime(df["date"])
```

```python
df["weekday"] = df["date"].dt.day_name()
df
```

```python
df["unit_price"] = df["price"] / df["quantity"]
```
