```python
import pandas as pd
from jedi.inference.names import AbstractNameDefinition
```

```python
data = {
    "date": ["01-01-2022", "01-01-2022", "02-01-2022", "03-01-2022", "03-01-2022"],
    "product": ["apple", "banana", "apple", "orange", "banana"],
    "price": [1.2, 0.8, 1.3, 1.5, 0.9],
    "quantity": [10, 5, 6, 8, 4],
    "city": ["Pahang", "Pahang", "Melaka", "Johor", "Pahang"]
}
```

```python
# & - AND
# | - or
# ~ - NOT
```

```python
# Price greater than 1.0 AND product apple
df[(df["price"] > 1.0) & (df["product"] == "apple")]
```

```python
# Price less than 1.0 OR quantity greater than 6
df[(df["price"] < 1.0) | (df["quantity"] > 6)]
```

```python
# NOT apple
df[~(df["product"] == "apple")]
```

```python
df.loc[df["city"] == "Pahang"]
```

```python
df.loc[0]
```

```python
 # df.loc[r,c]
df.loc[df["city"] == "Pahang", ["product","price"]]
```

```python
# Multi-index
multi = df.set_index(["city", "product"])
multi
```

```python
multi.loc[("Pahang", "apple")]
```

```python
# Exercise 1 - Select the rows where quantity is greater than 5 and show date and product only

df.loc[df["quantity"] > 5, ["date", "product"]]
```

```python
# Exercise 2 - Use iloc to select rows 1 through 3 and columns 0 through 3

df.iloc[1:4, 0:3]
```

```python
# Exercise 3 - Filter rows where city is Pahang OR product is orange

df[(df["city"] == "Pahang") | (df["product"] == "orange")]
```

```python
# Exercise 4 - Create a mask for price less than 1.0 and use it to filter rows

mask = df["price"] < 1.0
df[mask]
```

```python
# Exercise 5 - Convert df to a multi-index using date and city, then select all rows for 01-01-2022

mi = df.set_index(["date", "city"])
mi
```

```python
mi.loc[("01-01-2022", "Pahang")]
```
