```python
data = {
    "date": ["2024-01-01", "2024-01-02","2024/01/03", 'Jan 4 2024'],
    "sales": [100,120,90,160]
}
```

```python
df = pd.DataFrame(data)
df
```

```python
df["date"] = pd.to_datetime(df["date"], format="mixed")
```

```python
rng = pd.date_range(start="2024-01-01", periods=7, freq="D")
rng
```

```python
df = pd.DataFrame({"sales": [10,20,15,30,40,35,50]}, index=rng)
df
```

```python
df.resample("W").sum()
```

```python
df.resample("2D").mean()
```

```python
df
```

```python
df["rollind_3day"] = df["sales"].rolling(3).mean()
```

```python
df
```

```python
df["roll_sum_2day"] = df["sales"].rolling(2).sum()
```

```python
df
```

```python
df
```

```python
df["yesterday_sales"] = df["sales"].shift(1)
df
```

```python
df["change"] = df["sales"] - df["yesterday_sales"]
df
```

```python
df["shift_2"] = df["sales"].shift(2)
df
```

```python
df = df["sales"]
```

```python
df.index.day_name()
```

```python
df["month"] = df.index.month
```

```python
df
```

```python
df = df.drop("month")

# Convert Series to DataFrame
df = df.to_frame(name="sales")

# Ensure the index is datetime
df.index = pd.to_datetime(df.index)

# Add month and weekday columns
df["month"] = df.index.month
df["weekday"] = df.index.day_name()

df
```
